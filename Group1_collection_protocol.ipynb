## 1. Purpose

This protocol documents how the Asia, North America, and Europe pilot datasets were collected for Track 4. It explains the sampling frame, the attempted automated approach, the manual fallback, the coding rules, and the steps needed to reproduce the dataset.

### 2.2 Programme inclusion criteria
- Master’s-level programmes.
- Related to **Business Analytics, Data Science, Analytics, Applied Analytics, Artificial Intelligence, Applied Data Science, Statistics and Data Science, or Analytics and Data Science**.
- Active or listed for **2026** intake.
- Core curriculum covers **at least two** of the following five categories:
  - statistics
  - programming
  - machine learning
  - data visualisation
  - business analytics
- Full-time, on-campus mode.
- Official programme page available with enough information to code.

### 2.3 Time point
- **2026 only.**  
  Reliable 2021 admissions and curriculum evidence was not available, so this is a cross-sectional study. No claims about change over time are made.

### 2.4 Observation unit
- Final analysis: **programme-level**.

## 3. Data Sources

- Official university programme pages.
- Official programme PDFs where available.
- QS rankings used only to build the initial university list.
- No login, no CAPTCHA bypass, no scraping behind authentication.

---

## 4. Attempted Automated Collection

### 4.1 First attempt
- Used `requests` and `BeautifulSoup` to fetch programme pages and extract fields such as tuition, curriculum, and prerequisites.
- Checked `robots.txt` before scraping.
- The same automated-collection approach was considered/applied to the North America and Europe programme pages before manual verification.

### 4.2 Result
- **Asia**
  - **NTU** pages: static, `requests` could retrieve content.
  - **HKU, NUS** pages: dynamic and/or protected by WAF.  
    - HKU returned a generic template page; tuition was not in the raw HTML.
    - NUS returned a 212-byte empty response, indicating WAF blocking.
- **North America**
  - Programme information was not consistently available in a single static page and required manual verification across official programme/admission pages.
- **Europe**
  - Programme information was frequently distributed across separate official programme, curriculum, admissions, tuition, and English-requirement pages, so a single automated extraction rule was not reliable.
- Automated extraction **failed or was insufficiently reliable** for most target fields (tuition, curriculum, prerequisites) across the three regional samples.

### 4.3 Decision
- Switch to **manual collection** for Asia, North America, and Europe programmes.
- Use browser to open each official page, read fields, and record them into a coding sheet.

---

## 5. Manual Collection Procedure

For each university in the regional samples:

1. Go to the official graduate programme page.
2. Search for BA/DS/Analytics-related master’s programmes.
3. For each eligible programme:
   - Open the **2026** programme page.
   - Read the following fields:
     - programme name
     - university
     - country/region
     - school type
     - college type
     - ranking tier
     - core curriculum categories
     - prerequisites
     - GRE/GMAT policy
     - tuition amount and currency
     - duration in months
     - outcomes reported
     - English requirement
     - work experience requirement
     - application fee
     - minimum GPA
     - non-business background accepted
     - study mode
     - application deadline
   - Record the exact **source URL** and **access date**.
   - If a field is not available, write `NR`. Do not guess.
4. Save the row in the corresponding regional coding sheet.

---

## 7. Missing Data Handling

- Use `NR` for “Not Reported”.
- Never infer or fabricate values.
- Document missing fields in the README.
- If a programme does not meet the inclusion criteria (e.g. fewer than two core categories), exclude it and note the reason.

---

## 8. Quality Control

- Spot-check at least 10% of rows against the official page.
- Cross-check tuition and duration with a second source if possible (e.g. PDF brochure).
- If two coders are available, calculate inter-coder agreement on a sample.
- Record any disagreements and how they were resolved.

---

## 9. Tools Used

- Browser (Chrome)
- Excel for coding sheet
- `requests` and `BeautifulSoup` for attempted automated collection
- Selenium considered but not used for all pages due to time constraints

---

## 10. Ethical Considerations

- Checked `robots.txt` before any automated attempt.
- No login-walled or CAPTCHA-protected pages were accessed.
- No personal data was collected.
- Requests were low-frequency and identified with a User-Agent.
- When automated access was refused (e.g. WAF), manual collection was used instead of circumventing.
