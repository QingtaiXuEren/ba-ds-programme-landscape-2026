```python
from ddgs import DDGS
import pandas as pd
import time

def search_programmes(university, search_term, max_results=5):
    query = f'"{university}" "{search_term}" MSc OR Master'
    
    try:
        results = list(
            DDGS().text(
                query,
                max_results=max_results
            )
        )
        
        rows = []
        
        for result in results:
            rows.append({
                "university": university,
                "search_term": search_term,
                "title": result.get("title"),
                "url": result.get("href"),
                "snippet": result.get("body")
            })
        
        return rows
    
    except Exception as e:
        print(f"Error: {university} | {search_term} | {e}")
        return []
```

```python
MASTER_KEYWORDS = [
    "msc",
    "master",
    "masters",
    "postgraduate"
]

PROGRAMME_KEYWORDS = [
    "business analytics",
    "data science",
    "data analytics",
    "applied analytics",
    "business intelligence",
    "applied data science",
    "statistics and data science",
    "analytics and data science"
]

def looks_like_programme(title):
    title = str(title).lower()
    
    has_master = any(
        word in title
        for word in MASTER_KEYWORDS
    )
    
    has_programme = any(
        word in title
        for word in PROGRAMME_KEYWORDS
    )
    
    return has_master and has_programme
```

```python
from urllib.parse import urlparse
from ddgs import DDGS

EXCLUDED_DOMAINS = [
    "wikipedia.org",
    "topuniversities.com",
    "linkedin.com",
    "facebook.com",
    "instagram.com",
    "youtube.com",
    "timeshighereducation.com",
    "usnews.com",
    "mastersportal.com",
    "findamasters.com",
    "prospects.ac.uk",
    "ucas.com"
]

def find_official_domain(university):
    
    query = f'"{university}" official website'
    
    try:
        results = list(
            DDGS().text(
                query,
                max_results=10
            )
        )
        
        for result in results:
            
            url = result.get("href", "")
            
            if not url:
                continue
            
            domain = urlparse(url).netloc.lower()
            domain = domain.replace("www.", "")
            
            # Skip known third-party websites
            if any(
                excluded in domain
                for excluded in EXCLUDED_DOMAINS
            ):
                continue
            
            return domain
        
        return None
    
    except Exception as e:
        print(f"Error: {university} | {e}")
        return None
   
```

```python
from ddgs import DDGS
from urllib.parse import urlparse
import time

def search_official_programmes(
    university,
    official_domain,
    search_term,
    max_results=5
):
    
    query = f"{university} {search_term} MSc Master"
    
    rows = []
    
    try:
        results = list(
            DDGS().text(
                query,
                max_results=max_results
            )
        )
        
        for rank, result in enumerate(results, start=1):
            
            url = result.get("href", "")
            
            if not url:
                continue
            
            result_domain = (
                urlparse(url)
                .netloc
                .lower()
                .replace("www.", "")
            )
            
            # Keep only university official-domain results
            if (
                result_domain == official_domain
                or result_domain.endswith("." + official_domain)
            ):
                
                rows.append({
                    "university": university,
                    "search_term": search_term,
                    "query": query,
                    "result_rank": rank,
                    "title": result.get("title"),
                    "url": url,
                    "result_domain": result_domain,
                    "snippet": result.get("body")
                })
    
    except Exception as e:
        print(
            f"Error: {university} | "
            f"{search_term} | {e}"
        )
    
    return rows
```

### Extract Text from Accessible 2026 Programme Pages

```python
def extract_page_text(url):

    result = fetch_page(url)

    if result["status_code"] != 200:
        return None

    soup = BeautifulSoup(
        result["html"],
        "html.parser"
    )

    # Remove non-content elements
    for tag in soup([
        "script",
        "style",
        "nav",
        "footer",
        "header",
        "noscript"
    ]):
        tag.decompose()

    text = soup.get_text(
        separator=" ",
        strip=True
    )

    # Clean excessive whitespace
    text = " ".join(text.split())

    return text
```

```python
def find_keyword_context(
    text,
    keywords,
    context_chars=180
):
    if not text:
        return None

    text_lower = text.lower()
    matches = []

    for keyword in keywords:
        start = 0

        while True:
            pos = text_lower.find(
                keyword.lower(),
                start
            )

            if pos == -1:
                break

            left = max(
                0,
                pos - context_chars
            )

            right = min(
                len(text),
                pos + len(keyword) + context_chars
            )

            snippet = text[left:right]

            matches.append(
                f"[{keyword}] {snippet}"
            )

            start = pos + len(keyword)

            # 每个关键词最多保留3个结果
            if sum(
                m.startswith(f"[{keyword}]")
                for m in matches
            ) >= 3:
                break

    if not matches:
        return None

    return " || ".join(matches)
```

```python
def search_supporting_pages(
    university,
    programme_name,
    official_domain,
    page_type,
    query_terms,
    max_results=5
):
    query = (
        f'"{programme_name}" '
        f'{query_terms} '
        f'site:{official_domain}'
    )

    rows = []

    try:
        results = list(
            DDGS().text(
                query,
                max_results=max_results
            )
        )

        for rank, result in enumerate(
            results,
            start=1
        ):
            url = result.get("href", "")

            if not url:
                continue

            result_domain = (
                urlparse(url)
                .netloc
                .lower()
                .replace("www.", "")
            )

            if (
                result_domain == official_domain
                or result_domain.endswith(
                    "." + official_domain
                )
            ):
                rows.append({
                    "page_type": page_type,
                    "result_rank": rank,
                    "title": result.get("title"),
                    "url": url,
                    "snippet": result.get("body")
                })

    except Exception as e:
        print(
            f"Error | {university} | "
            f"{page_type} | {e}"
        )

    return rows
```

```python
import re

DURATION_PATTERNS = [
    r'(\d+)\s*months?',
    r'(\d+)\s*month',
    r'(\d+)\s*years?',
    r'(\d+)\s*year',
]

def find_duration_evidence(text):

    if not text:
        return None

    matches = []

    for pattern in DURATION_PATTERNS:

        for match in re.finditer(
            pattern,
            text,
            flags=re.IGNORECASE
        ):

            start = max(
                0,
                match.start() - 120
            )

            end = min(
                len(text),
                match.end() + 120
            )

            matches.append(
                text[start:end]
            )

            if len(matches) >= 5:
                break

        if len(matches) >= 5:
            break

    if not matches:
        return None

    return " || ".join(matches)
```

```python
import requests
import pandas as pd
import time
from bs4 import BeautifulSoup

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 Chrome/120 Safari/537.36"
    )
}

def fetch_page_text(url):
    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=15
        )

        status_code = response.status_code

        if status_code != 200:
            return {
                "status_code": status_code,
                "page_text": pd.NA,
                "text_length": 0,
                "fetch_status": "Failed"
            }

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        # Remove non-content elements
        for tag in soup(
            [
                "script",
                "style",
                "noscript",
                "svg"
            ]
        ):
            tag.decompose()

        text = " ".join(
            soup.stripped_strings
        )

        return {
            "status_code": status_code,
            "page_text": text,
            "text_length": len(text),
            "fetch_status": "Success"
        }

    except Exception as e:
        return {
            "status_code": pd.NA,
            "page_text": pd.NA,
            "text_length": 0,
            "fetch_status": f"Error: {str(e)[:80]}"
        }
```

```python
import re
import pandas as pd


EVIDENCE_KEYWORDS = {
    "prerequisites": [
        "entry requirements",
        "admission requirements",
        "admissions requirements",
        "bachelor",
        "bachelor's degree",
        "undergraduate degree",
        "prerequisite",
        "requirements profile"
    ],

    "GRE_GMAT": [
        "GRE",
        "GMAT",
        "Graduate Record Examination",
        "Graduate Management Admission Test"
    ],

    "tuition": [
        "tuition",
        "tuition fee",
        "tuition fees",
        "fees",
        "overseas tuition",
        "international tuition"
    ],

    "duration": [
        "duration",
        "calendar year",
        "months",
        "years",
        "ECTS"
    ],

    "statistics": [
        "statistics",
        "statistical",
        "probability",
        "regression",
        "econometrics"
    ],

    "programming": [
        "programming",
        "Python",
        "R programming",
        "coding",
        "computer programming"
    ],

    "machine_learning": [
        "machine learning",
        "deep learning",
        "artificial intelligence",
        "predictive analytics"
    ],

    "visualization": [
        "data visualization",
        "data visualisation",
        "visualization",
        "visualisation"
    ],

    "business_analytics": [
        "business analytics",
        "business intelligence",
        "decision analytics",
        "analytics for business"
    ],

    "outcomes": [
        "employment",
        "employed",
        "graduate outcomes",
        "career outcomes",
        "graduate destinations",
        "employers"
    ]
}


def extract_context(text, keywords, window=220):
    if pd.isna(text):
        return pd.NA

    text = str(text)

    matches = []

    for keyword in keywords:
        pattern = re.compile(
            re.escape(keyword),
            flags=re.IGNORECASE
        )

        match = pattern.search(text)

        if match:
            start = max(
                0,
                match.start() - window
            )

            end = min(
                len(text),
                match.end() + window
            )

            snippet = text[start:end]

            snippet = re.sub(
                r"\s+",
                " ",
                snippet
            ).strip()

            matches.append(snippet)

    if not matches:
        return pd.NA

    # Remove duplicate snippets
    unique_matches = list(
        dict.fromkeys(matches)
    )

    return " || ".join(
        unique_matches[:3]
    )
```

```python
import re
import pandas as pd


def extract_duration_candidate(text):
    if pd.isna(text):
        return pd.NA

    text = str(text)

    patterns = [
        # 12 months / 18 months / 24 months
        r'\b(12|18|24)\s*months?\b',

        # 1 year / 2 years
        r'\b(1|2)\s*(?:calendar\s+)?years?\b'
    ]

    candidates = []

    for pattern in patterns:
        for match in re.finditer(
            pattern,
            text,
            flags=re.IGNORECASE
        ):
            value = int(match.group(1))

            if "year" in match.group(0).lower():
                value = value * 12

            candidates.append(value)

    if not candidates:
        return pd.NA

    return " | ".join(
        map(
            str,
            sorted(set(candidates))
        )
    )


programme_review[
    "duration_candidate"
] = programme_review[
    "duration_evidence"
].apply(
    extract_duration_candidate
)
```

```python
def extract_tuition_candidates(text):
    if pd.isna(text):
        return pd.NA

    text = str(text)

    patterns = [
        r'£\s?[\d,]+',
        r'€\s?[\d,.]+',
        r'CHF\s?[\d,]+',
        r'DKK\s?[\d,.]+',
        r'EUR\s?[\d,.]+',
        r'GBP\s?[\d,.]+'
    ]

    candidates = []

    for pattern in patterns:
        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        candidates.extend(matches)

    # remove duplicates
    candidates = list(
        dict.fromkeys(candidates)
    )

    if not candidates:
        return pd.NA

    return " | ".join(
        candidates[:10]
    )


programme_review[
    "tuition_candidate"
] = programme_review[
    "tuition_evidence"
].apply(
    extract_tuition_candidates
)
```

```python
def classify_test_policy_candidate(text):
    if pd.isna(text):
        return "Not reported"

    text_lower = str(text).lower()

    not_required_patterns = [
        "gre is not required",
        "gre not required",
        "gmat is not required",
        "gmat not required",
        "not require the gre",
        "not require gre",
        "not require gmat"
    ]

    required_patterns = [
        "gre is required",
        "gre required",
        "gmat is required",
        "gmat required"
    ]

    optional_patterns = [
        "gre optional",
        "gmat optional",
        "gre is optional",
        "gmat is optional"
    ]

    if any(
        x in text_lower
        for x in not_required_patterns
    ):
        return "Possible: Not required"

    if any(
        x in text_lower
        for x in required_patterns
    ):
        return "Possible: Required"

    if any(
        x in text_lower
        for x in optional_patterns
    ):
        return "Possible: Optional"

    # GRE/GMAT appeared but context is unclear
    if (
        "gre" in text_lower
        or "gmat" in text_lower
    ):
        return "Mentioned - manual review"

    return "Not reported"


programme_review[
    "GRE_GMAT_candidate"
] = programme_review[
    "GRE_GMAT_evidence"
].apply(
    classify_test_policy_candidate
)
```

```python
def extract_duration_sentences(text):
    if pd.isna(text):
        return pd.NA

    text = re.sub(r"\s+", " ", str(text))

    # Split roughly into sentences / page fragments
    pieces = re.split(
        r'(?<=[.!?])\s+|(?=\bDuration\b)|(?=\bLength\b)',
        text
    )

    duration_patterns = [
        r'\b\d+\s*(?:calendar\s+)?years?\b',
        r'\b\d+\s*months?\b',
        r'\b\d+\s*ECTS\b',
        r'\bfull[- ]time\b',
        r'\bpart[- ]time\b'
    ]

    hits = []

    for piece in pieces:
        piece_lower = piece.lower()

        # Prefer pieces actually related to programme duration
        duration_context = any(
            word in piece_lower
            for word in [
                "duration",
                "full-time",
                "full time",
                "part-time",
                "part time",
                "programme",
                "program",
                "study",
                "ects"
            ]
        )

        has_duration_value = any(
            re.search(
                pattern,
                piece,
                flags=re.IGNORECASE
            )
            for pattern in duration_patterns[:3]
        )

        if duration_context and has_duration_value:
            hits.append(
                piece.strip()[:600]
            )

    hits = list(dict.fromkeys(hits))

    if not hits:
        return pd.NA

    return " || ".join(hits[:5])
```

```python
def extract_tuition_sentences(text):
    if pd.isna(text):
        return pd.NA

    text = re.sub(r"\s+", " ", str(text))

    # Split page into manageable pieces
    pieces = re.split(
        r'(?<=[.!?])\s+|(?=\bTuition\b)|(?=\bFees?\b)',
        text
    )

    # Currency / amount patterns
    money_patterns = [
        r'£\s?[\d,]+',
        r'€\s?[\d,.]+',
        r'CHF\s?[\d,]+',
        r'DKK\s?[\d,.]+',
        r'EUR\s?[\d,.]+',
        r'GBP\s?[\d,.]+'
    ]

    hits = []

    for piece in pieces:

        piece_lower = piece.lower()

        # Must be related to tuition / fees
        fee_context = any(
            word in piece_lower
            for word in [
                "tuition",
                "tuition fee",
                "tuition fees",
                "fee",
                "fees",
                "international",
                "overseas",
                "non-eu",
                "non-eu/eea",
                "institutional fee"
            ]
        )

        # Must contain a monetary amount
        has_money = any(
            re.search(
                pattern,
                piece,
                flags=re.IGNORECASE
            )
            for pattern in money_patterns
        )

        if fee_context and has_money:
            hits.append(
                piece.strip()[:800]
            )

    hits = list(dict.fromkeys(hits))

    if not hits:
        return pd.NA

    return " || ".join(hits[:6])
```

```python
import requests
import time

def find_2021_snapshot(url):
    cdx_url = "https://web.archive.org/cdx/search/cdx"

    params = {
        "url": url,
        "from": "2021",
        "to": "2021",
        "output": "json",
        "filter": "statuscode:200",
        "filter": "mimetype:text/html",
        "collapse": "digest",
        "fl": "timestamp,original,statuscode",
        "limit": "1"
    }

    try:
        r = requests.get(cdx_url, params=params, timeout=20)

        if r.status_code == 200:
            data = r.json()

            if len(data) > 1:
                timestamp = data[1][0]
                original = data[1][1]

                archive_url = (
                    f"https://web.archive.org/web/"
                    f"{timestamp}/{original}"
                )

                return 1, archive_url

        return 0, pd.NA

    except Exception:
        return pd.NA, pd.NA


for i, row in historical_check.iterrows():

    available, archive_url = find_2021_snapshot(
        row["official_url"]
    )

    historical_check.loc[
        i, "archive_available_2021"
    ] = available

    historical_check.loc[
        i, "archive_url_2021"
    ] = archive_url

    time.sleep(0.5)


display(
    historical_check[
        [
            "programme_id",
            "university",
            "programme_name",
            "archive_available_2021",
            "archive_url_2021"
        ]
    ]
)

print(
    "\n2021 snapshot found:",
    (historical_check["archive_available_2021"] == 1).sum()
)

print(
    "No snapshot for current URL:",
    (historical_check["archive_available_2021"] == 0).sum()
)

print(
    "Request/error unresolved:",
    historical_check["archive_available_2021"].isna().sum()
)
```
